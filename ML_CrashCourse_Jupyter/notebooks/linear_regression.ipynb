# Linear Regression

## Prediction Equation

The prediction of a linear regression model is:

$$
y' = b + w_1x_1
$$

Where:

- \(y'\) = predicted value
- \(b\) = bias (intercept)
- \(w_1\) = model weight
- \(x_1\) = feature value

### Linear regression examples and graphs

#### Import linear regression variables

```python
import numpy as np
import matplotlib.pyplot as plt
```

#### 1.Basic linear regression graph

```python
# Model parameters
b = 3
w = 5

# Generate x values
x = np.linspace(-5, 5, 100)

# Linear regression prediction
y = b + w * x

# Plot
plt.figure(figsize=(8, 5))
plt.plot(x, y, label=r"$y' = b + wx$")
plt.xlabel("Feature (x)")
plt.ylabel("Prediction ($y'$)")
plt.title("Linear Regression Prediction")
plt.grid(True)
plt.legend()
plt.show()
```

#### 2. Show how changing the weight changes the slope

```python

b = 1
x = np.linspace(-5, 5, 100)

for w in [0.5, 1, 2, 3]:
    y = b + w * x
    plt.plot(x, y, label=f"w = {w}")

plt.xlabel("x")
plt.ylabel("Prediction")
plt.title("Effect of Changing the Weight")
plt.grid(True)
plt.legend()
plt.show()
```

#### 3. Show how changing the bias shifts the line

```python
w = 1
x = np.linspace(-5, 5, 100)

for b in [0.5, 1, 2, 3]:
    y = b + w * x
    plt.plot(x, y, label=f"b = {b}")

plt.xlabel("x")
plt.ylabel("Prediction")
plt.title("Effect of Changing the bias")
plt.grid(True)
plt.legend()
plt.show()
```

#### 4. Show data points and best fit line

```python
# Example training data
x = np.array([1, 2, 3, 4, 5])
y = np.array([5, 7, 8, 11, 12])

# Model
b = 2
w = 2

# Predictions
y_pred = b + w * x

plt.figure(figsize=(8, 5))
plt.scatter(x, y, label="Training data")
plt.plot(x, y_pred, label="Model prediction")
plt.xlabel("Feature")
plt.ylabel("Label")
plt.title("Linear Regression")
plt.legend()
plt.grid(True)
plt.show()
```
