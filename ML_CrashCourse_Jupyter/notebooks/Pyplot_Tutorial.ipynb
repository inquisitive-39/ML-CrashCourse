Get data from dataset online and plot it out. This example shows how pyplot works with dataframes to read values from a column even though the variable passed is the column name which is a string. Normally, you have to parse an array

```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

df = pd.read_csv(url)

# View the data
print(df.head())

# Scatter plot
plt.figure(figsize=(8, 6))
plt.scatter(
    "sepal_length",
    "petal_length",
    data=df
)

# The above is interchangeable with 
# plt.figure(figsize=(8, 6))
# plt.scatter(
#     df["sepal_length"],
#     df["petal_length"]
# )

plt.title("Iris Dataset")
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")

plt.show()
```

This code generates a synthetic dataset containing four variables: a is a sequence of numbers from 0 to 49, b is a with added random noise to create a loose correlation, c contains random values used to determine the colour of each point, and d contains random positive values used to determine point size. It then creates a scatter plot where a and b control the position of each point on the graph, while c and d control the appearance of the points. The strings passed to plt.scatter() are automatically matched to variables in the data dictionary via the data=data parameter. The result is a visualisation that displays four dimensions of data in a single chart.

```python
data = {'a': np.arange(50),
        'c': np.random.randint(0, 50, 50),
        'd': np.random.randn(50)}
data['b'] = data['a'] + 10 * np.random.randn(50)
data['d'] = np.abs(data['d']) * 100

plt.scatter('a', 'b', c='c', s='d', data=data)
plt.xlabel('entry a')
plt.ylabel('entry b')
plt.show()
```

This code defines a function f(t) that produces a cosine wave whose amplitude gradually decreases over time due to the exponential decay factor exp(-t). It then creates two arrays of time values: t1 with fewer points (every 0.1 units) and t2 with more points (every 0.02 units). The figure is split into two subplots: the top subplot compares the same decaying cosine function using blue circle markers ('bo') for the coarse data (t1) and a smooth black line ('k') for the finer data (t2), demonstrating how more data points produce a smoother plot. The bottom subplot displays a regular cosine wave without decay using a red dashed line ('r--'), allowing you to compare a damped oscillation with an undamped one.

Following code shows how to work with figure (the window) and multiple axes (the plots/graphs)

Figure
│
├── Axes (Graph 1)
│
├── Axes (Graph 2)
│
└── Axes (Graph 3)


```python
import numpy as np
import matplotlib.pyplot as plt

# Define a function that returns a decaying cosine wave
def f(t):
    return np.exp(-t) * np.cos(2 * np.pi * t)

# Create time values from 0 to 5 in steps of 0.1 (fewer points)
t1 = np.arange(0.0, 5.0, 0.1)

# Create time values from 0 to 5 in steps of 0.02 (more points)
t2 = np.arange(0.0, 5.0, 0.02)

# Create a new figure (canvas)
plt.figure()

# Select the first subplot (2 rows, 1 column, position 1)
plt.subplot(211)

# Plot the decaying cosine wave:
# 'bo' = blue circle markers using t1 (coarse data)
# 'k' = black line using t2 (smooth data)
plt.plot(t1, f(t1), 'bo', t2, f(t2), 'k')

# Select the second subplot (2 rows, 1 column, position 2)
plt.subplot(212)

# Plot a regular cosine wave:
# 'r--' = red dashed line
plt.plot(t2, np.cos(2 * np.pi * t2), 'r--')

# Display the figure containing both subplots
plt.show()
```
