# Linear Regression

## Prediction Equation

The prediction of a linear regression model is:

$$
y' = b + w_1x_1
$$

Where:

- \(y'\) = predicted value
- \(b\) = bias (intercept)
- \(w_1\) = model weight
- \(x_1\) = feature value
